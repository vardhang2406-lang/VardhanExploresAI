```python
import pandas as pd
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
df = pd.read_csv("/Users/vardhangaur/alzheimer's_ml/data.csv")
```

```python
df
```

```python
df['ID'] = df['ID'].str.replace('id_', '').astype(int)
```

```python
df_new = df.rename(columns={'ID':'Patient_Number'})
```

```python
df_new.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_new['class']=le.fit_transform(df_new['class'])
```

```python
df_new
```

```python
df_new.corr()
```

```python
X = df_new.drop(columns='class')
y = df_new[['class']]
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
```

```python
param_grid = {
    'n_estimators': [50, 100, 200],      
    'max_depth': [None, 5, 10],          
    'min_samples_split': [2, 5],
    'criterion':['gini','entropy']        
}
```

```python
grid_search = GridSearchCV(
    estimator=model,        # The blank model
    param_grid=param_grid,     # The 18 combinations
    cv=5,                      # 5-fold cross-validation
    scoring='accuracy',        # Grade them based on accuracy
    n_jobs=-1                  # Run tests in parallel using all computer cores
)

grid_search.fit(X_train, y_train)
print("Search complete!")
```

```python
best_model=grid_search.best_estimator_
```

```python
final_accuracy = best_model.score(X_test, y_test)
```

```python
y_pred = best_model.predict(X_test)
accuracy = accuracy_score(y_pred,y_test)
print(accuracy)
```

```python
best_params =grid_search.best_params_
```
